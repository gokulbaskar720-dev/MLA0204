```python

import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

# Dataset
data = {
    'Outlook': ['sunny', 'sunny', 'overcast', 'rain', 'rain',
                'rain', 'overcast', 'sunny', 'sunny', 'rain',
                'sunny', 'overcast', 'overcast', 'rain'],

    'Temperature': ['hot', 'hot', 'hot', 'mild', 'cool',
                    'cool', 'cool', 'mild', 'cool', 'mild',
                    'mild', 'mild', 'hot', 'mild'],

    'Humidity': ['high', 'high', 'high', 'high', 'normal',
                 'normal', 'normal', 'high', 'normal', 'normal',
                 'normal', 'high', 'normal', 'high'],

    'Wind': ['weak', 'strong', 'weak', 'weak', 'weak',
             'strong', 'strong', 'weak', 'weak', 'weak',
             'strong', 'strong', 'weak', 'strong'],

    'Target': ['no', 'no', 'yes', 'yes', 'yes',
               'no', 'yes', 'no', 'yes', 'yes',
               'yes', 'yes', 'yes', 'no']
}

df = pd.DataFrame(data)

# Separate input features and target
X = pd.get_dummies(df[['Outlook', 'Temperature', 'Humidity', 'Wind']])
y = df['Target']

# Build ID3 decision tree using entropy
model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, y)

# Display decision tree
print("Decision Tree:")
print(export_text(model, feature_names=list(X.columns)))

# Classify a new sample
sample = pd.DataFrame([{
    'Outlook': 'sunny',
    'Temperature': 'cool',
    'Humidity': 'normal',
    'Wind': 'strong'
}])

sample_encoded = pd.get_dummies(sample).reindex(
    columns=X.columns, fill_value=0
)

prediction = model.predict(sample_encoded)

print("\nNew Sample:")
print(sample)

print("Predicted Target:", prediction[0])
```
