```python

# Candidate-Elimination Algorithm

training_data = [
    (['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same'], 'Yes'),
    (['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same'], 'Yes'),
    (['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change'], 'No'),
    (['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change'], 'Yes')
]

S = ['Ø'] * 6
G = [['?'] * 6]

domains = [
    ['Sunny', 'Rainy'],
    ['Warm', 'Cold'],
    ['Normal', 'High'],
    ['Strong', 'Weak'],
    ['Warm', 'Cool'],
    ['Same', 'Change']
]

def covers(h, x):
    return all(a == '?' or a == b for a, b in zip(h, x))

def more_general(h1, h2):
    return all(a == '?' or a == b for a, b in zip(h1, h2))

def generalize(h, x):
    result = h.copy()
    for i in range(len(h)):
        if result[i] == 'Ø':
            result[i] = x[i]
        elif result[i] != x[i]:
            result[i] = '?'
    return result

def specialize(h, x):
    result = []
    for i in range(len(h)):
        if h[i] == '?':
            for value in domains[i]:
                if value != x[i]:
                    new_h = h.copy()
                    new_h[i] = value
                    result.append(new_h)
    return result

for x, label in training_data:
    if label == 'Yes':
        G = [g for g in G if covers(g, x)]
        S = x.copy() if 'Ø' in S else generalize(S, x)
        G = [g for g in G if more_general(g, S)]
    else:
        new_G = []
        for g in G:
            if covers(g, x):
                new_G.extend(specialize(g, x))
            else:
                new_G.append(g)
        G = [list(h) for h in {tuple(g) for g in new_G}]
        G = [g for g in G if more_general(g, S)]

    print("\nExample:", x, label)
    print("S =", S)
    print("G =", G)

print("\nFinal Specific Boundary (S):", S)
print("Final General Boundary (G):")
for h in G:
    print(h)
```
